## V. HƯỚNG DẪN CHẠY BẰNG DÒNG LỆNH (CLI & ORCHESTRATION)

Toàn bộ pipeline có thể chạy độc lập không cần Jupyter thông qua file `run_pipeline.py`:

```bash

python run_pipeline.py --mode monitor
```
