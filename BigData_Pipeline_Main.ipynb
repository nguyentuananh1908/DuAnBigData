## Pipeline Summary

### Data Source
- **Source:** E-commerce user behavior data (view, cart, purchase)
- **Format:** CSV files (2019-Oct to 2020-Feb)
- **Size:** ~5 months of data

````python
# Quick stats
from pathlib import Path

data_dir = Path('data')
if data_dir.exists():
    files = list(data_dir.glob('*.csv'))
    print('📁 Data Files:')
    for f in files:
        size_mb = f.stat().st_size / (1024*1024)
        print(f'   {f.name}: {size_mb:.1f} MB')
else:
    print('Data folder not found. Create data/ folder and add CSV files.')
````

### Processing Strategy

| Aspect | Choice | Reason |
|--------|--------|--------|
| **Mode** | Batch | Data size manageable, daily processing sufficient |
| **ETL/ELT** | ETL | Clean data before storage, simpler architecture |
| **Tools** | Python/Pandas | Sufficient for dataset size |
| **Storage** | CSV + PostgreSQL | Flexible, scalable |
| **ML** | Sklearn | Simple, effective |
| **Monitoring** | Custom | Lightweight, meets requirements |
