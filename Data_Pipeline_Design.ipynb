## 3. Xử lý Dirty Data

### 3.1 Missing Values
- `category_code` = NULL → Unknown
- `brand` = "" → Unknown

### 3.2 Inconsistent Categories
- Lowercase tất cả
- Remove dấu cách thừa

### 3.3 Invalid Sessions
- Remove NULL sessions
- Filter duration = 0

### 3.4 Outliers
- IQR cho price
- Validate timestamp range
