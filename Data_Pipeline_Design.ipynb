## 3. Xử lý Dirty Data

### 3.1 Missing Values
- `category_code` = NULL → Unknown
- `brand` = "" → Unknown

### 3.2 Inconsistent Categories
- Lowercase tất cả
- Remove dấu cách thừa

### 3.3 Invalid Sessions
- Remove NULL sessions
- Filter duration = 0

### 3.4 Outliers
- IQR cho price
- Validate timestamp range

````python
# Vi du ETL
import pandas as pd

# Doc CSV
df = pd.read_csv('data/2019-Oct.csv')

# Clean data
df['brand'] = df['brand'].fillna('Unknown')
df['category_code'] = df['category_code'].fillna('Unknown')

# Remove outliers
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)
IQR = Q3 - Q1
df = df[(df['price'] >= Q1 - 1.5*IQR) & (df['price'] <= Q3 + 1.5*IQR)]

print(f'Shape after cleaning: {df.shape}')
````
